```python
# Task 1: Import NumPy, Pandas, Matplotlib and Seaborn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Load train.csv into a Pandas DataFrame
df = pd.read_csv("/Titanic-Dataset.csv")

```

```python
# Task 3: Display the first 10 rows
df.head(10)
```

```python
# Task 4: Display the last 5 rows

df.tail(5)
```

```python
# Task 5: Check dataset dimensions

df.shape
```

```python
# Task 6: Display dataset information

df.info()
```

```python
# Task 7: Generate summary statistics

df.describe()
```

```python
# Task 7: Display summary statistics for Age and Fare

df[["Age", "Fare"]].describe()
```

```python
# Task 8: Identify missing values

df.isnull().sum()
```

```python
# Task 9: Fill missing Age values with the median

df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
# Task 9: Verify that missing Age values have been filled

df["Age"].isnull().sum()
```

```python
# Task 10: Filter surviving passengers where Survived == 1

survived_passengers = df[df["Survived"] == 1]

survived_passengers
```

```python
# Task 11: Convert Fare column to a NumPy array

fare_array = df["Fare"].to_numpy()

fare_array
```

```python
# Task 12: Calculate maximum and average Fare

max_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", max_fare)
print("Average Fare:", average_fare)
```

```python
# Task 13: Plot histogram of Age

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20)

plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Histogram of Age")

plt.show()
```

```python
# Task 14: Plot bar chart of Survival Count

survival_count = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(6, 5))

plt.bar(["Did Not Survive", "Survived"], survival_count)

plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Survival Count")

plt.show()
```

```python
# Task 15: Plot scatter graph of Age vs Fare

plt.figure(figsize=(8, 5))

plt.scatter(df["Age"], df["Fare"])

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")

plt.show()
```
